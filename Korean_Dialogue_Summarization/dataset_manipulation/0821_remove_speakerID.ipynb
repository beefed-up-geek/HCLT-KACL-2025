```python
# ==== 1. 파일 경로 지정 ====
abs = "/Users/taeyoonkwack/Documents/HCLT-KACL-2025/Korean_Dialogue_Summarization/"
file = abs+"dataset/original_english_formatted/train.json"
INPUT_JSON = file 
OUTPUT_JSON = abs+"dataset/original_english_formatted/temp.json"
```

```python
import json

# ==========================
# 🔄 변환 함수
# ==========================
def replace_speakers(data):
    new_data = []
    for item in data:
        # speaker_map: {"화자1": {"id": "...", "name": "..."}, "화자2": {...}}
        speaker_map = item.get("speaker_map", {})

        # 매핑 테이블 생성 (id, name -> 화자명)
        replacement_dict = {}
        simplified_map = {}
        for speaker, info in speaker_map.items():
            speaker_id = info.get("id", "")
            replacement_dict[speaker_id] = speaker
            replacement_dict[info.get("name", "")] = speaker
            simplified_map[speaker] = speaker_id  # id만 남김

        # dialogue와 output에서 교체
        dialogue = item["dialogue"]
        output = item["output"]

        for old, new in replacement_dict.items():
            if old:  # 빈 문자열 방지
                dialogue = dialogue.replace(old, new)
                output = output.replace(old, new)

        # 새로운 아이템 구성
        new_item = {
            "id": item["id"],
            "dialogue": dialogue,
            "subject_keyword": item["subject_keyword"],
            "speaker_map": simplified_map,
            "output": output
        }
        new_data.append(new_item)

    return new_data

# ==========================
# 🚀 실행
# ==========================
with open(INPUT_JSON, "r", encoding="utf-8") as f:
    data = json.load(f)

converted_data = replace_speakers(data)

with open(OUTPUT_JSON, "w", encoding="utf-8") as f:
    json.dump(converted_data, f, ensure_ascii=False, indent=2)

print(f"✅ 변환 완료! 저장 위치: {OUTPUT_JSON}")
```
