```python
import json

# 입력 및 출력 파일 경로 설정
input_path = "../dataset/original_english_formatted/dev.json"
output_path = "../dataset/original_english_formatted/dev.json"

# JSON 파일 읽기
with open(input_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 각 항목 처리
for item in data:
    if "output_en" in item:
        item["output"] = item["output_en"]  # output_en 내용 덮어쓰기
        del item["output_en"]               # output_en 삭제

    if "subject_keyword_en" in item:
        item["subject_keyword"] = item["subject_keyword_en"]  # subject_keyword_en 내용 덮어쓰기
        del item["subject_keyword_en"]                        # subject_keyword_en 삭제

# 수정된 데이터 저장
with open(output_path, 'w', encoding='utf-8') as f:
    json.dump(data, f, ensure_ascii=False, indent=2)

print(f"변환 완료: {output_path}")
```
