```python
import json

# 입력 / 출력 파일 경로
input_file = "../dataset/original/dev.json"     # 원본 JSON
output_file = "../dataset/original_formatted/dev.json"  # 변환 후 저장할 JSON

# JSON 파일 읽기
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

converted_data = []

for item in data:
    dialogue_lines = []
    
    # utterance를 화자1 / 화자2로 변환
    for utt in item["input"]["utterance"]:
        speaker = "화자1" if utt["speaker_id"] == "P1" else "화자2"
        dialogue_lines.append(f"{speaker}: {utt['form']}")
    
    # output 그대로 유지
    output_labels = item["output"]
    
    converted_data.append({
        "id": item["id"],
        "dialogue": "\n".join(dialogue_lines),
        "output": output_labels
    })

# 변환 결과 저장
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(converted_data, f, ensure_ascii=False, indent=2)

print(f"변환 완료. 저장 위치: {output_file}")
```

```python
import json

# 입력 / 출력 파일 경로
input_file = "../dataset/original/test.json"     # 원본 JSON
output_file = "../dataset/original_formatted/test.json"  # 변환 후 저장할 JSON

# JSON 읽기
with open(input_file, "r", encoding="utf-8") as f:
    data = json.load(f)

converted_data = []

for item in data:
    dialogue_lines = []
    
    # utterance 변환 및 output 생성
    output_labels = []
    for utt in item["input"]["utterance"]:
        speaker = "화자1" if utt["speaker_id"] == "P1" else "화자2"
        dialogue_lines.append(f"{speaker}: {utt['form']}")
        
        # output에 추가
        output_labels.append({
            "id": utt["id"],
            "label": "inappropriate"
        })
    
    converted_data.append({
        "id": item["id"],
        "dialogue": "\n".join(dialogue_lines),
        "output": output_labels
    })

# 저장
with open(output_file, "w", encoding="utf-8") as f:
    json.dump(converted_data, f, ensure_ascii=False, indent=2)

print(f"변환 완료. 저장 위치: {output_file}")
```
