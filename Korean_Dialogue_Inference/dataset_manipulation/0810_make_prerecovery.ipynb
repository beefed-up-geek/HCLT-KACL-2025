```python
import json

input_file = "../dataset/original/test.json"       # 입력 JSON 파일
output_file = "../dataset/original_formatted/test.json"  # 출력 JSON 파일

# ====== 1. 데이터 불러오기 ======
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
def convert_entry(entry):
    entry_id = entry["id"]

    # 대화 변환
    dialogue_lines = []
    for turn in entry["input"]["conversation"]:
        dialogue_lines.append(f"화자{turn['speaker']}: {turn['utterance']}")
    dialogue_text = "\n".join(dialogue_lines)

    # question 변환
    q_text = entry["input"]["question"].strip()
    choices = [
        f"A. {entry['input']['inference_1']}",
        f"B. {entry['input']['inference_2']}",
        f"C. {entry['input']['inference_3']}",
    ]
    question_text = q_text + "\n" + "\n".join(choices)

    # answer 변환
    inference_map = {
        "inference_1": "A",
        "inference_2": "B",
        "inference_3": "C"
    }
    answer_letter = inference_map[entry["output"]]

    return {
        "id": entry_id,
        "dialogue": dialogue_text,
        "question": question_text,
        "answer": answer_letter
    }
```

```python
converted_data = [convert_entry(e) for e in data]
```

```python
with open(output_file, 'w', encoding='utf-8') as f:
    json.dump(converted_data, f, ensure_ascii=False, indent=2)

print(f"변환 완료! JSON 저장 위치: {output_file}")
```
